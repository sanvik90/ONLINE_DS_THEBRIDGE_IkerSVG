## Ejercicio 2
Escribe un programa para pasar de grados a radianes. Hay que usar `input`. Recuerda que la conversión se realiza mediante

radianes = grados*(pi/180)

```python
import math  
pi = 3.141592653589793
grados = float(input("Bienvendio al convertidor de Grados en Radianes.Por favor introduzca los grados que desea convertir"))
decimales = int(input("Por favor establezca el número de decimiales que desea"))
radianes = round(grados *(pi/180),decimales) # Para acortar vamos a pedir 6 dcimales.

print(grados,"º grados Equivalen a",radianes,"radianes. \nGracias por haber usado nuestro converitdor.")




# El paquete de math.pi tiene una funcion para calcular radianes. Vamos a usarla para comprobar que nuestro programa ha funcionado.
print("_________________________________________ \n"
"Validador con math.pi`")
if radianes ==  round(math.radians(grados),decimales):
    print("Math.pi da un valor de", round(math.radians(grados),decimales)," por ello el convertidor ha funcionado satisfactoriamente")
else:
    print("El convertidor no ha funcioando correctamente")

```

## Ejercicio 3
Escribe un programa que calcule el area de un paralelogramo (base x altura). También con `input`

```python

atierra = 5100720000000000000
base = int(input("Bienvienido al calculador de areas de paralelogramos The Bridge Max543-F. \nPor favor,introduzca la altura de su paralelogramo en centímetros, sin decimales"))
altura = int(input("Por favor, hagalo ahora con la altura. También en centímetros,sin decimales"))
if base <= 0  or altura<= 0:
    print("Recorcholis, \nsu paralelogramo desafía las leyes de la física conocidas. \nPor favor,pongase en contacto con el departamento de físca experimental de su unviersidad más cercana.") 
else:
    area = base * altura
    if area >= atierra:
        print("Impresionante. \nSu paralelogramo tiene un área mayor que la del planeta tierra.\nSu area es de", area,"cm cuadrados")
    else:
        if base == altura:
            print("Le informamos de que su paralelogramo es un cuadrado,\ny su area es de ", area,"cm cuadrados. ")
        else:
            print("El area de su paralelogramo es de", area,"cm cuadrados ")
print("Gracias por haber usado el calculador de areas de paralelogramos The Bridge Max543-F.¡Hasta la próxima!")

#Los saltos e ifelses permiten  que el programa deje de correr cuando hay alguna variable negativa o se exceda el area de la tierra.
```

## Ejercicio 4
Escribe un programa que calcule el area de un paralelogramo (base x altura). También con `input`. En este caso debe ser capaz de admitir valores reales, con decimales, para base y altura en español. (PISTA: Recuerda que tendrás que hacer uso del método replace)

```python
# A la hora de pedir decimales el usario Europeo puede encontrar errores y meter la coma
base = input("Bienvienido al calculador de areas de paralelogramos The Bridge Max900-J ProDecimal. \nPor favor,introduzca la altura de su paralelogramo en centímetros.")
altura = input("Hagalo ahora con la alutura")
numdecimal = int(input("Cuantos decimales quiere cómo máximo"))

#Ahora usamos replace para todos aquellos usarios Europeaos que saben que en en el mundo real los decimales van despues de una coma
base = base.replace(",",".")

altura = altura.replace(",",".")

# Ahora que ya hemos solucionado el tema de los puntos y comas pasamos las variables a floats.
base = float(base)
altura = float(altura)

#Una vez con las variables convertidas en floats podemos replicar el programa anterior con una ligera modificación
# que permite elegir el número de decimales del area
if base <= 0  or altura<= 0:
    print("Recorcholis, \nsu paralelogramo desafía las leyes de la física conocidas. \nPor favor,pongase en contacto con el departamento de físca experimental de su unviersidad más cercana.") 
else:
    area = base * altura
    if area >= atierra:
        print("Impresionante. \nSu paralelogramo tiene un área mayor que la del planeta tierra.\nSu area es de", round(area,numdecimal),"cm cuadrados")
    else:
        if base == altura:
            print("Le informamos de que su paralelogramo es un cuadrado,\ny su area es de ", round(area,numdecimal),"cm cuadrados. ")
        else:
            print("El area de su paralelogramo es de", round(area,numdecimal),"cm cuadrados ")
print("Gracias por haber usado el calculador de areas de paralelogramos The Bridge Max900-J ProDecimal¡Hasta la próxima!")
```
