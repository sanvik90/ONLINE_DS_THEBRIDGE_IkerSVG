## Ejercicio 1
Tenemos las siguientes variables:

```Python
A = 4
B = "Text"
C = 4.1
```

Comprueba:
1. Si A y B son equivalentes
2. Si A y C NO son equivalentes
3. Si A es mayor que C
4. Si C es menor o igual que A
5. Si B NO es equivalente a C

````python
#Variables  
A =4
B = "Text"
C = 4.1
#proposiciones 
pre1 = A == C
pre2 =  A != C
pre3 = A >=C
pre4 =  C <= A or C==A
pre5 =  B != C
#Outputs
print("La proposición 1:'A y B son equivalentes' es", pre1,".")
print("La proposición 2: 'A y C NO son equivalentes' es", pre2)
print("La proposición 3: 'A es mayor que C' es", pre3)
print("La proposición 4: 'C es menor o igual que A'", pre4)
print("La proposición 5: B NO es equivalente a C", pre5)
````

````python
print("Bienvenido al comprobador de equivalencias de dos inputs. \n Por favor, siga usted las instruccione")
input1 = input("Introduzca el primer input")
input2 = input("Introduzca el segundo input")

resultado = input1 == input2

if resultado ==  True:
    print("¡Felicidades! \n Ambos inputs son equivalentes")
else:
    print("Los dos inputs NO! son equivalentes")
````
