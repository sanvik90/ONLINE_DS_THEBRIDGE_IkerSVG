## Ejercicio 2
En este ejercicio vamos a implementar una clase *Perro* en Python. La clase tiene las siguientes características:
* Cosas que sabemos seguro que tiene un perro
    * Tiene 4 patas
    * 2 orejas
    * 2 ojos
    * Una velocidad de 0. Por defecto, el perro está parado
* Cuando se inicialice:
    * El perro será de una determinada raza
    * Por defecto tendrá pelo "Marrón", a no ser que se diga lo contrario.
    * Por defecto no tendrá dueño, a no ser que se diga lo contrario.
    
* Dispondrá también de un método llamado andar, que tiene un argumento de entrada (aumento_velocidad). Este valor se le sumará a la velocidad que ya llevaba el perro.
* Necesita otro método (parar), donde pondremos la velocidad a 0.
* Otro método llamado "ladrar", que tendrá un argumento de entrada, y la salida será el siguiente string: "GUAU!" + el argumento de entrada.


Se pide:
* Implementa la clase *Perro*
* Crea un objeto de tipo *Perro*, sin dueño
* Comprueba que están bien todos sus atributos
* Prueba que ande, y comprueba su velocidad
* Páralo
* Documenta la clase *Perro*

```python
class Perro():
    Perros = []
    "Clase que recoje las caracteristicas de diferentes perros"
    patas = 4
    orejas = 2
    ojos = 2 
    vel = 0
    def __init__(self, raza, pelo="Marrón", dueño=None):
        "Introducir la raza del perro. Es insispensable"
        self.raza = raza
        "Por defecto se asigna el color marrón"
        self.pelo = pelo
        "Por defecto se asocia ningun perro"
        self.dueño = dueño
        Perro.Perros.append(self)

    def andar(self,aumento_velocidad):
        "Andar ofrece una velocidad al perro en fucnión del aumento de velocidaddado"
        self.vel = self.vel + aumento_velocidad
        return print(f"Ahora la velocidad de su perro es de {self.vel}")
    def parar(self):
        "Parar hace que su perro se par"
        self.vel = 0
        return print("¡ Su perro se ha parado !")
    def ladrar(self,ladrido):
        "ladrar hace que su perro ladre el argumento dado"
        return print(f"GUAU!" + ladrido )
```

```python
Firu1 = Perro("Border Collie","Blanco y negro")
print(Firu1.patas)
print(Firu1.orejas)
print(Firu1.ojos)
print(Firu1.vel)
print(Firu1.raza)
print(Firu1.pelo)
print(Firu1.dueño)
```

```python
aumento_velocidad = int(input("Introduzca el aumento de velocidad para el perro"))
ladrido = str(input("¿Qué va  a ladrar su perro"))
Firu1.andar(aumento_velocidad )
Firu1.parar()
print(Firu1.vel)
Firu1.ladrar(ladrido)
```
